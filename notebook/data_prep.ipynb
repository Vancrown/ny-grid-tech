### concat data


```python
df1 = pl.read_csv("../data/archive/nyiso_lmp_day_ahead_hourly_2025.csv")
df2 = pl.read_csv("../data/archive/nyiso_lmp_day_ahead_hourly_0101_0331.csv")
df3 = pl.read_csv("../data/archive/nyiso_lmp_day_ahead_hourly_0401_0403.csv")
```

```python
df = pl.concat([df1, df2, df3], how="vertical").filter(
    pl.col("location_type") == "Zone"
)
```

```python
df.write_csv("../data/nyiso_lmp_day_ahead_hourly.csv")
```
