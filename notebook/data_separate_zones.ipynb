```python
import pandas as pd
from pathlib import Path
```

```python
# Input file
input_csv = "../data/nyiso_lmp_day_ahead_hourly.csv"   # change this to your actual file path
```

```python
# Output folder
output_dir = Path("../Agentics_Energy/agentic_energy/data/NYISO_zones")
output_dir.mkdir(parents=True, exist_ok=True)
```

```python
# Read the CSV
df = pd.read_csv(input_csv)
```

```python
# Keep the columns you want
cols = [
    "interval_start_local",
    "location",
    "lmp",
    "energy",
    "congestion",
    "loss",
]

# LMP = Energy + Loss + Congestion
```

```python
df = df[cols].copy()

# Rename interval_start_local -> timestamps
df = df.rename(columns={"interval_start_local": "timestamps"})

# Rename lmp -> price, since on this final price we do arbitrage
df = df.rename(columns={"lmp": "prices"})

# Create one CSV per zone
for zone, zone_df in df.groupby("location"):
    safe_zone = (
        zone.replace(" ", "_")
            .replace(".", "")
            .replace("/", "_")
    )
    zone_df = zone_df.drop(columns=["location"])
    output_path = output_dir / f"{safe_zone}.csv"
    zone_df.to_csv(output_path, index=False)
    print(f"Saved: {output_path}")
```
